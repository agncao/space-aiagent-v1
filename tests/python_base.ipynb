```python
from contextvars import ContextVar

var = ContextVar('var')
token  =var.set(1)
print(var.get())
var.set(2)
print(var.get())
var.reset( token)
# print(var.get())  # 表示取得token之前的值，但是token之前没有赋值所以此时打印出来是异常
```

```python
import asyncio


async def func(task_name: str):
    print(f'{task_name} start')
    await asyncio.sleep(0)
    print(f'{task_name} end')

tasks = [asyncio.ensure_future(func('task_name'+i) for i in range(10))]

loop = asyncio.get_event_loop()
loop.run_until_complete(asyncio.wait(tasks))
```
